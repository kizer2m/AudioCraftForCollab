### Alternative: Direct Python API Usage (MusicGen Example)
If you prefer to generate music via code directly inside notebook cells:

```python
import torch
from audiocraft.models import MusicGen
from audiocraft.data.audio import audio_write
from IPython.display import Audio, display

# 1. Load MusicGen small model
model = MusicGen.get_pretrained('facebook/musicgen-small')
model.set_generation_params(duration=10) # 10 seconds

# 2. Generate audio from text prompt
descriptions = ['synthwave track with heavy bass, energetic drums, and 80s synth lead']
print('Generating music...')
wav = model.generate(descriptions) # [1, 1, sample_rate * duration]

# 3. Save to WAV file
audio_data = wav[0].cpu()
audio_write('generated_sample', audio_data, model.sample_rate, strategy='loudness', add_suffix=False)

# 4. Play audio in notebook
display(Audio('generated_sample.wav'))
```
